# Notebook 02: Transaction Graph Analysis & Temporal Partitioning

Inspect graph structure, feature distributions, node degree imbalance, and verify leak-free temporal splitting.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / "src"))

from adamule.data.dataset import TransactionGraphDataset
from adamule.visualization.graph_plots import plot_subgraph
import matplotlib.pyplot as plt

dataset = TransactionGraphDataset(profile="development")
graph = dataset.process()

print("Number of nodes:", graph.num_nodes)
print("Number of edges:", graph.num_edges)
print("Train active nodes:", graph.train_mask.sum().item())
print("Val active nodes:", graph.val_mask.sum().item())
print("Test active nodes:", graph.test_mask.sum().item())

# Visualize ego-network around first fraud node
fraud_nodes = (graph.y == 1).nonzero(as_tuple=True)[0].numpy()
fig = plot_subgraph(graph, center_node=int(fraud_nodes[0]), hops=2)
plt.show()
```
