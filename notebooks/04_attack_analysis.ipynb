# Notebook 04: Adversarial Structuring & Camouflage Evasion

Evaluate fraudster evasion through financial constraint validation, amount smurfing, and merchant camouflage.

```python
import sys
from pathlib import Path
import torch
sys.path.insert(0, str(Path.cwd().parent / "src"))

from adamule.data.dataset import TransactionGraphDataset
from adamule.models.gcn import GCN
from adamule.attacks.heuristic import HeuristicAttacker
from adamule.visualization.attack_plots import plot_attack_trajectory
import matplotlib.pyplot as plt

dataset = TransactionGraphDataset(profile="development")
graph = dataset.process()

model = GCN(in_dim=graph.x.shape[1], hidden_dim=64)
attacker = HeuristicAttacker(detector=model)

fraud_nodes = (graph.y == 1).nonzero(as_tuple=True)[0].numpy()
target = int(fraud_nodes[0])
evasive_graph, summary = attacker.attack_node(graph, target_node=target, max_steps=8)
print("Attack Summary:", summary)

fig = plot_attack_trajectory(summary["history"], target_node=target)
plt.show()
```
