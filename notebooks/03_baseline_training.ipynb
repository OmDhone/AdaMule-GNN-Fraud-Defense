# Notebook 03: Baseline Graph Fraud Detectors Training

Train and compare baseline architectures: GCN, GAT, and CARE-GNN.

```python
import sys
from pathlib import Path
import torch
sys.path.insert(0, str(Path.cwd().parent / "src"))

from adamule.data.dataset import TransactionGraphDataset
from adamule.models.gcn import GCN
from adamule.training.trainer import GNNTrainer

dataset = TransactionGraphDataset(profile="development")
graph = dataset.process()

model = GCN(in_dim=graph.x.shape[1], hidden_dim=64)
opt = torch.optim.Adam(model.parameters(), lr=0.005)
trainer = GNNTrainer(model, opt)
results = trainer.fit(graph, epochs=20)
print("Test Metrics:", results["test_metrics"])
```
